```python
#step 5
#Convert to Datetime and extracting components
superstore['Order Date'] = pd.to_datetime(superstore['Order Date'])
superstore['Ship Date'] = pd.to_datetime(superstore['Ship Date'])

superstore['Order Day'] = superstore['Order Date'].dt.day
superstore['Order Month'] = superstore['Order Date'].dt.month
superstore['Order Year'] = superstore['Order Date'].dt.year

superstore['Ship Day'] = superstore['Ship Date'].dt.day
superstore['Ship Month'] = superstore['Ship Date'].dt.month
superstore['Ship Year'] = superstore['Ship Date'].dt.year
superstore
```

```python
#step 7
#Detecting Outliers: IQR Rule: Outliers = Q1 - 1.5IQR or Q3 + 1.5IQR

numeric_columns = superstore.select_dtypes(include='number').columns
superstore[numeric_columns].describe()

Q1 = superstore["Sales"].quantile(0.25)
Q3 = superstore["Sales"].quantile(0.75)
IQR = Q3 - Q1
print(IQR)

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

sales_outliers = superstore[
    (superstore['Sales'] < lower_bound) |
    (superstore['Sales'] > upper_bound)
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("sales_outliers:", sales_outliers)
```

```python
Q1 = superstore["Quantity"].quantile(0.25)
Q3 = superstore["Quantity"].quantile(0.75)
IQR = Q3 - Q1
print(IQR)

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

quantity_outliers = superstore[
    (superstore['Quantity'] < lower_bound) |
    (superstore['Quantity'] > upper_bound)
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("quantity_outliers:", quantity_outliers)
```

```python
Q1 = superstore['Profit'].quantile(0.25)
Q3 = superstore['Profit'].quantile(0.75)

IQR = Q3 - Q1
print(IQR)

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

profit_outliers = superstore[
    (superstore['Profit'] < lower_bound) |
    (superstore['Profit'] > upper_bound)
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("profit_outliers:", profit_outliers)
```

```python
#step 8
# Create a copy of the original dataset
cleaned_superstore = superstore.copy()

# Remove exact duplicate rows
cleaned_superstore = cleaned_superstore.drop_duplicates()

# Convert date columns to datetime
cleaned_superstore['Order Date'] = pd.to_datetime(
    cleaned_superstore['Order Date'], errors='coerce'
)

cleaned_superstore['Ship Date'] = pd.to_datetime(
    cleaned_superstore['Ship Date'], errors='coerce'
)

# Create date-related columns
cleaned_superstore['Order Day'] = cleaned_superstore['Order Date'].dt.day
cleaned_superstore['Order Month'] = cleaned_superstore['Order Date'].dt.month
cleaned_superstore['Order Year'] = cleaned_superstore['Order Date'].dt.year

cleaned_superstore['Ship Day'] = cleaned_superstore['Ship Date'].dt.day
cleaned_superstore['Ship Month'] = cleaned_superstore['Ship Date'].dt.month
cleaned_superstore['Ship Year'] = cleaned_superstore['Ship Date'].dt.year

# Check the cleaned dataset
print("Shape:", cleaned_superstore.shape)

print("\nMissing values:")
print(cleaned_superstore.isnull().sum())

print("\nData types:")
print(cleaned_superstore.dtypes)
```
